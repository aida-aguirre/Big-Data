### Explicación del modelo ERD

**1.Tabla Customer**

- customer_id: Clave primaria que identifica de manera única a cada cliente.

- country: País del cliente, útil para análisis geográfico y segmentación de mercado.

Descripción:
Esta tabla almacena la información de los clientes que realizaron compras. Sirve como base para relacionar las ventas con su ubicación geográfica y analizar comportamientos de compra por región.

**2.Tabla Product**

- product_id: Clave primaria que identifica de manera única cada producto.

- product_name: Nombre o descripción del producto.

- unit_price: Precio unitario del producto, necesario para calcular ingresos y márgenes.

Descripción:
Contiene la información de los productos disponibles en la tienda online.
Permite relacionar las transacciones con los artículos vendidos y analizar la demanda por producto o categoría.

**3.Tabla Sales**

- transaction_id: Clave primaria que identifica cada transacción de venta.

- customer_id: Clave foránea que se relaciona con Customer.customer_id.

- product_id: Clave foránea que se relaciona con Product.product_id.

- quantity: Cantidad de unidades vendidas en la transacción.

- date: Fecha en que se realizó la venta.

Descripción:
Registra cada operación comercial, asociando un cliente con un producto específico en una fecha determinada.
Esta tabla es el núcleo del modelo, ya que permite realizar análisis de ventas, ticket promedio, comportamiento de clientes y rendimiento de productos.

### Relaciones

- **Customer → Sales:** Un cliente puede realizar muchas ventas (`1:N`), pero cada venta corresponde a un solo cliente.  
- **Product → Sales:** Un producto puede aparecer en muchas ventas (`1:N`), y cada venta contiene un producto específico.  

Este modelo relacional refleja la estructura del dataset “An Online Shop Business”.
Asegura que los datos estén organizados de manera lógica y consistente, estableciendo la base para la creación de la base de datos real y la ejecución de consultas analíticas a futuro.

## 3️⃣ Creación de la base de datos y carga de información

Se creó la base de datos SQLite tienda_online.db con tres tablas: Customer (customer_id, country), Product (product_id, product_name, unit_price) y Sales (transaction_id, date, product_id, customer_id, quantity). Los datos del dataset se cargaron en pandas, se limpiaron valores nulos, se renombraron las columnas para coincidir con el modelo relacional y se insertaron en SQLite usando to_sql(). Finalmente, se verificó la correcta carga de la información consultando los primeros registros de cada tabla y realizando consultas de ejemplo para confirmar las relaciones entre clientes, productos y ventas.

### Instalación de librerías y descarga del dataset

```python
# Librerías necesarias
import os
import json
import zipfile
from pathlib import Path
from typing import List, Any

import numpy as np
import pandas as pd
import kagglehub

# Descargar dataset desde Kaggle
dataset_path = kagglehub.dataset_download("gabrielramos87/an-online-shop-business")
print("Ruta a los archivos del dataset:", dataset_path)
```

### Cargar el dataset como DataFrame de pandas

```python
# Revisar si hay archivos zip, csv o xlsx en la ruta descargada
archivos = os.listdir(dataset_path)
print("Archivos descargados:", archivos)

# Si hay zip, extraer
zip_files = [f for f in archivos if f.endswith(".zip")]
if zip_files:
    zip_file = os.path.join(dataset_path, zip_files[0])
    extract_dir = os.path.join(dataset_path, "extracted")
    os.makedirs(extract_dir, exist_ok=True)
    with zipfile.ZipFile(zip_file, "r") as z:
        z.extractall(extract_dir)
    data_dir = extract_dir
else:
    # Si no hay zip, usar la ruta descargada directamente
    data_dir = dataset_path

print("Ruta final de datos:", data_dir)
```

```python
# Buscar archivos CSV o XLSX
archivos = os.listdir(data_dir)
csv_files = [f for f in archivos if f.endswith(".csv")]
xlsx_files = [f for f in archivos if f.endswith(".xlsx")]

# Cargar CSV
if csv_files:
    df = pd.read_csv(os.path.join(data_dir, csv_files[0]), encoding="latin1")
elif xlsx_files:
    df = pd.read_excel(os.path.join(data_dir, xlsx_files[0]))
else:
    raise FileNotFoundError("No se encontró archivo CSV ni XLSX en el dataset")

# Mostrar las primeras filas
df.head()
```

### Limpieza básica de datos

```python
# Revisar valores nulos
df.isnull().sum()

# Reemplazar nulos
df = df.fillna({
    "CustomerNo": 0,
    "ProductNo": "ND",
    "ProductName": "ND",
    "Price": 0.0,
    "Country": "ND",
    "Date": "1900-01-01" # fecha por defecto si falta
})

# Renombrar columnas para profesionalizar el dataset
df.rename(columns={
    'TransactionNo': 'transaction_id',
    'Date': 'date',
    'ProductNo': 'product_id',
    'ProductName': 'product_name',
    'Price': 'unit_price',
    'Quantity': 'quantity',
    'CustomerNo': 'customer_id',
    'Country': 'country'
}, inplace=True)

print("✅ Columnas renombradas correctamente:")
print(df.columns.tolist())



```

### Creación de la Base de Datos e Inserción de Información en SQLite

```python
import sqlite3
import pandas as pd

# Crear o conectar a la base de datos local
conn = sqlite3.connect("tienda_online.db")

# Crear tablas basadas en el modelo ER profesional
df_customers = df[['customer_id', 'country']].drop_duplicates().copy()
df_products = df[['product_id', 'product_name', 'unit_price']].drop_duplicates().copy()
df_sales = df[['transaction_id', 'date', 'product_id', 'customer_id', 'quantity']].copy()

# Guardar las tablas en SQLite
df_customers.to_sql("Customer", conn, if_exists="replace", index=False)
df_products.to_sql("Product", conn, if_exists="replace", index=False)
df_sales.to_sql("Sales", conn, if_exists="replace", index=False)

print("✅ Base de datos 'tienda_online.db' creada exitosamente con nombres profesionalizados.")
```

```python
print("📋 Tablas en la base de datos:")
print(pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn))
```

```python
print("📋 Vista previa de cada tabla:")

print("\n🧍 Customer:")
display(pd.read_sql("SELECT * FROM Customer LIMIT 5;", conn))

print("\n📦 Product:")
display(pd.read_sql("SELECT * FROM Product LIMIT 5;", conn))

print("\n💳 Sales:")
display(pd.read_sql("SELECT * FROM Sales LIMIT 5;", conn))
```

**4. Consultas con filtros:** Para evidenciar la correcta relación entre tablas y la posibilidad de extraer información específica:

```python
### Consulta filtrada: Ventas de clientes en Noruega
query = """
SELECT DISTINCT s.transaction_id, s.date, s.quantity, p.product_name, c.country
FROM Sales s
JOIN Product p ON s.product_id = p.product_id
JOIN Customer c ON s.customer_id = c.customer_id
WHERE c.country = 'Norway'
LIMIT 10;
"""
display(pd.read_sql(query, conn))
```

Interpretación:

- Esta consulta muestra las ventas realizadas por clientes de Noruega, combinando información de las tablas Sales, Product y Customer.

- Se observa que una misma transacción (transaction_id = 536532) puede contener múltiples productos, lo que refleja la relación 1:N entre ventas y productos.

- Los datos incluyen la fecha de la venta, cantidad vendida, nombre del producto y país del cliente, evidenciando que las tablas están correctamente relacionadas.

- Esto confirma que la base de datos fue cargada correctamente y que es posible realizar filtros y análisis por país, cliente o producto.
